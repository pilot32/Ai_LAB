```python
#Set Creation and Set Length

f={'apple','oranges','grapes','litchis','sahil'}
v={'Potato','tomato','cucumber','dragon fruit','sahil'}
print(len(f))
print(len(v))
print(f)
print(v)
```

```python
#Allocation of Different Data Types into a Set

s={'taylor',36,89,+918917225803,'unmarried',False}
print(s)
s
```

```python
#Using set() constructor

s=set([1,2,3,4])
s
```

```python
#Accessing Set Items

s={'taylor',36,89,+918917225803,'unmarried',False}
for items in s:
    print(items)
```

```python
#Adding items to a Set using add() and updates()

s={'taylor',36,89,+918917225803,'unmarried',False}
s.add('sahil')
print(s)
```
